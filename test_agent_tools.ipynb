# ConversationAgent Tools Testing

This notebook demonstrates the tools available to the ConversationAgent:
1. **compute** - Mathematical/Statistical/Financial calculations
2. **jot** - Working memory for multi-hop reasoning

The agent uses ReAct pattern to intelligently decide when to use these tools.

## Setup

```python
import sys
from pathlib import Path

# Add src to path
sys.path.insert(0, str(Path.cwd() / "src"))

# Load environment variables
from dotenv import load_dotenv
load_dotenv()

from src.infrastructure.llm.conversation_agent import ConversationAgent
from src.application.interfaces.conversation_agent import (
    ConversationContext,
    Message,
)

# Initialize agent
agent = ConversationAgent(temperature=0.7)
```

## Test 1: compute Tool - Simple Arithmetic

```python
context = ConversationContext(
    current_message="What is 157 * 234 + 89?"
)

response = agent.generate_response(context)
print(f"Response: {response.content}\n")
if response.reasoning:
    print(f"Trajectory: {response}")
    print(f"Reasoning: {response.reasoning}")
```

## Test 5: Combined Tools - Complex Problem Solving

```python
context = ConversationContext(
    current_message="""A bakery sells cupcakes for $3 each and cookies for $2 each. 
    If they sold 45 cupcakes and 78 cookies yesterday, and their costs were $120 for ingredients, 
    what was their profit margin percentage?"""
)

response = agent.generate_response(context)
print(f"Response: {response.content}\n")
if response.reasoning:
    print(f"Reasoning: {response.reasoning}")
```

## Test 6: Multi-Turn Conversation with Context

```python
# First message
context1 = ConversationContext(
    current_message="Calculate the area of a circle with radius 5 meters."
)
response1 = agent.generate_response(context1)
print(f"Turn 1 - Response: {response1.content}\n")

# Second message with history
history = (
    Message(role="user", content="Calculate the area of a circle with radius 5 meters."),
    Message(role="assistant", content=response1.content),
)

context2 = ConversationContext(
    current_message="Now calculate the volume if this is the base of a cylinder with height 10 meters.",
    history=history
)
response2 = agent.generate_response(context2)
print(f"Turn 2 - Response: {response2.content}\n")
```
