# Expense Tracking Tools - Interactive Testing

This notebook demonstrates the expense tracking system with real examples.

**Setup**: Ensure you have:
- PostgreSQL with pgvector extension installed
- Database migration `002_create_expenses_table.sql` executed
- Gemini API key configured in `.env`

## 1. Setup and Configuration

```python
import asyncio
import json
from datetime import date
from decimal import Decimal

# Import necessary modules
from sqlalchemy.ext.asyncio import create_async_engine, async_sessionmaker
from src.config.settings import settings
from src.infrastructure.llm.embedding.gemini_embedding_service import GeminiEmbeddingService
from src.infrastructure.llm.tools.expenses import (
    configure_expense_tools,
    set_user_context,
    expense_tools
)

print("✅ Imports successful")
```

```python
# Create database engine and session factory
engine = create_async_engine(str(settings.database_url), echo=False)
session_factory = async_sessionmaker(engine, expire_on_commit=False)

# Initialize embedding service
embedding_service = GeminiEmbeddingService(
    api_key=settings.gemini_api_key,
    model_name=settings.embedding_model
)

# Configure expense tools
configure_expense_tools(
    session_factory=session_factory,
    embedding_service=embedding_service
)

# Set user context
set_user_context(user_id="notebook_test_user")

print("✅ Expense tools configured successfully")
print(f"📊 Database: {settings.database_url}")
print(f"👤 User: notebook_test_user")
```

## 2. Creating Expenses

Let's create various types of expenses to test the system.

```python
# Create a basic food expense
result1 = expense_tools(
    operation="create",
    amount=85.50,
    description="Lunch at Italian restaurant with colleagues",
    category="food",
    tags=["restaurant", "lunch", "italian"]
)
print("Result 1:", result1)

# Extract expense ID for later use
expense_id_1 = result1.split("ID: ")[1].split(",")[0] if "ID: " in result1 else None
print(f"\n💾 Saved expense ID: {expense_id_1}")
```

```python
# Create a recurring rent expense
result2 = expense_tools(
    operation="create",
    amount=15000,
    currency="HKD",
    description="Monthly rent payment for October",
    category="housing",
    tags=["rent", "recurring", "essential"],
    expense_date="2025-10-01",
    is_recurring=True,
    recurrence_pattern="monthly"
)
print("Result 2:", result2)

expense_id_2 = result2.split("ID: ")[1].split(",")[0] if "ID: " in result2 else None
print(f"\n💾 Saved expense ID: {expense_id_2}")
```

```python
# Create a transport expense
result3 = expense_tools(
    operation="create",
    amount=45.50,
    description="Taxi from office to client meeting at ABC Corp",
    category="transport",
    tags=["taxi", "client", "business"],
    expense_date="2025-10-25"
)
print("Result 3:", result3)

expense_id_3 = result3.split("ID: ")[1].split(",")[0] if "ID: " in result3 else None
print(f"\n💾 Saved expense ID: {expense_id_3}")
```

```python
# Create a subscription expense in USD
result4 = expense_tools(
    operation="create",
    amount=9.99,
    currency="USD",
    description="Netflix monthly subscription",
    category="entertainment",
    tags=["subscription", "streaming", "monthly"],
    is_recurring=True,
    recurrence_pattern="monthly"
)
print("Result 4:", result4)

expense_id_4 = result4.split("ID: ")[1].split(",")[0] if "ID: " in result4 else None
print(f"\n💾 Saved expense ID: {expense_id_4}")
```

```python
# Create more sample expenses for testing
sample_expenses = [
    {"amount": 120, "description": "Grocery shopping at supermarket", "category": "food", "tags": ["groceries", "essential"]},
    {"amount": 350, "description": "Dinner with clients at steakhouse", "category": "business-meals", "tags": ["client", "dinner", "reimbursable"]},
    {"amount": 25.50, "description": "Morning coffee and breakfast", "category": "food", "tags": ["coffee", "breakfast"]},
    {"amount": 80, "description": "Gym membership monthly fee", "category": "health", "tags": ["gym", "recurring"], "is_recurring": True, "recurrence_pattern": "monthly"},
    {"amount": 200, "description": "Office supplies from stationary store", "category": "office", "tags": ["office", "supplies", "reimbursable"]},
]

created_ids = []
for exp in sample_expenses:
    result = expense_tools(operation="create", **exp)
    print(result)
    if "ID: " in result:
        exp_id = result.split("ID: ")[1].split(",")[0]
        created_ids.append(exp_id)

print(f"\n✅ Created {len(created_ids)} additional expenses")
```

## 3. Reading and Listing Expenses

```python
# Get a single expense by ID
if expense_id_1:
    result = expense_tools(operation="get", expense_id=expense_id_1)
    print("Single Expense Details:")
    print(result)
    
    # Parse and pretty print JSON
    try:
        expense_data = json.loads(result)
        print("\n📋 Parsed Data:")
        print(f"  Amount: {expense_data['amount']} {expense_data['currency']}")
        print(f"  Description: {expense_data['description']}")
        print(f"  Category: {expense_data['category']}")
        print(f"  Tags: {', '.join(expense_data['tags'])}")
        print(f"  Date: {expense_data['expense_date']}")
    except:
        pass
```

```python
# List all expenses (default limit 10)
result = expense_tools(operation="list")
print("All Expenses:")
print(result)
```

```python
# List expenses with pagination
result = expense_tools(operation="list", limit=5, offset=0)
print("First 5 Expenses:")
print(result)
```

```python
# List expenses for a specific date range
result = expense_tools(
    operation="list",
    date_from="2025-10-01",
    date_to="2025-10-31",
    limit=20
)
print("October 2025 Expenses:")
print(result)
```

## 4. Updating Expenses

```python
# Update amount and add tags
if expense_id_1:
    result = expense_tools(
        operation="update",
        expense_id=expense_id_1,
        amount=95.00,
        tags=["restaurant", "lunch", "italian", "team"]
    )
    print("Update Result:", result)
    
    # Verify the update
    updated = expense_tools(operation="get", expense_id=expense_id_1)
    print("\nUpdated Expense:")
    print(updated)
```

```python
# Update description (this will regenerate the embedding)
if expense_id_3:
    result = expense_tools(
        operation="update",
        expense_id=expense_id_3,
        description="Taxi ride from office to important client meeting at ABC Corporation headquarters"
    )
    print("Update Result:", result)
```

## 5. Searching Expenses

Test different search types to find expenses.

```python
# Semantic search - find expenses by meaning
result = expense_tools(
    operation="search",
    query="where did I eat out?",
    search_type="semantic",
    limit=5
)
print("Semantic Search - Dining Expenses:")
print(result)
```

```python
# Semantic search - client meetings
result = expense_tools(
    operation="search",
    query="client meeting expenses",
    search_type="semantic",
    limit=5
)
print("Semantic Search - Client Meetings:")
print(result)
```

```python
# Category search - find all food expenses
result = expense_tools(
    operation="search",
    search_type="category",
    category="food",
    limit=10
)
print("Category Search - Food:")
print(result)
```

```python
# Tag search - find reimbursable expenses
result = expense_tools(
    operation="search",
    search_type="tags",
    tags=["reimbursable"],
    limit=10
)
print("Tag Search - Reimbursable:")
print(result)
```

```python
# Tag search - recurring expenses
result = expense_tools(
    operation="search",
    search_type="tags",
    tags=["recurring", "monthly"],
    limit=10
)
print("Tag Search - Recurring/Monthly:")
print(result)
```

```python
# Amount range search - expensive items (over 100 HKD)
result = expense_tools(
    operation="search",
    search_type="amount_range",
    amount_min=100,
    limit=10
)
print("Amount Range Search - Over 100 HKD:")
print(result)
```

```python
# Amount range search - small expenses (under 50 HKD)
result = expense_tools(
    operation="search",
    search_type="amount_range",
    amount_max=50,
    limit=10
)
print("Amount Range Search - Under 50 HKD:")
print(result)
```

```python
# Amount range search - mid-range expenses (50-200 HKD)
result = expense_tools(
    operation="search",
    search_type="amount_range",
    amount_min=50,
    amount_max=200,
    limit=10
)
print("Amount Range Search - Between 50-200 HKD:")
print(result)
```

Test error handling with invalid inputs.

```python
# Test negative amount
result = expense_tools(
    operation="create",
    amount=-50,
    description="Test",
    category="test"
)
print("Negative Amount Error:")
print(result)
```

```python
# Test invalid currency code
result = expense_tools(
    operation="create",
    amount=100,
    currency="US",  # Too short
    description="Test",
    category="test"
)
print("Invalid Currency Error:")
print(result)
```

```python
# Test invalid recurrence pattern
result = expense_tools(
    operation="create",
    amount=100,
    description="Test",
    category="test",
    is_recurring=True,
    recurrence_pattern="hourly"  # Not supported
)
print("Invalid Recurrence Pattern Error:")
print(result)
```

## 9. Advanced Usage Examples

```python
# Example: Find all business expenses that need reimbursement
print("=" * 60)
print("BUSINESS EXPENSE REIMBURSEMENT REPORT")
print("=" * 60)

# Search by tag
reimbursable = expense_tools(
    operation="search",
    search_type="tags",
    tags=["reimbursable"],
    limit=50
)
print(reimbursable)

# Get totals
totals = expense_tools(
    operation="analytics",
    analytics_type="total_by_category",
    date_from="2025-10-01",
    date_to="2025-10-31"
)
print("\n" + totals)
```

```python
# Example: Budget monitoring - check if over budget in any category
print("=" * 60)
print("BUDGET MONITORING REPORT")
print("=" * 60)

# Define monthly budgets
budgets = {
    "food": 2000,
    "transport": 500,
    "entertainment": 800,
    "housing": 15000,
    "health": 1000,
}

# Get monthly summary
summary = expense_tools(
    operation="analytics",
    analytics_type="monthly_summary",
    year=2025,
    month=10
)
print(summary)

print("\n💰 Budget vs Actual:")
for category, budget in budgets.items():
    # Search category total
    result = expense_tools(
        operation="search",
        search_type="category",
        category=category,
        limit=100
    )
    if "No expenses found" not in result:
        print(f"  {category.title()}: Budget {budget} HKD")
```

```python
# Example: Track recurring expenses
print("=" * 60)
print("RECURRING EXPENSES SUMMARY")
print("=" * 60)

recurring = expense_tools(
    operation="search",
    search_type="tags",
    tags=["recurring"],
    limit=50
)
print(recurring)
```
